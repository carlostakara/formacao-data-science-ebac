Faça a leitura dos dados do arquivo CSV:

```python
df = pd.read_csv("MODULO7_PROJETOFINAL_BASE_SUPERMERCADO.csv", delimiter=',')

df.head(10)
```

Os campos do nosso dataframe são:

**Title:** Nome do produto.


**Marca:** A marca do produto.


**Preco_Normal:** O preço em que o produto costuma ser vendido quando não há desconto.


**Preco_Desconto:** O preço vendido após o desconto ser aplicado.


**Preco_Anterior:** Preço em que era comercializado o produto antes do desconto aplicado.


**Desconto:** Total de desconto aplicado.






As colunas que aparecem com valores 0 são para os produtos onde não tivemos descontos aplicados.


As categorias estão em espanhol!

# 1 - Traga a média e a mediana dos preços - coluna Preco_Normal - por categoria de produto.
# Identifique as categorias que parecem ter um valor de média abaixo ou acima da mediana.

```python
media_categoria = df.groupby('Categoria')['Preco_Normal'].mean().sort_values(ascending=False)
print(media_categoria)
```

```python
mediana_categoria = df.groupby('Categoria')['Preco_Normal'].median().sort_values(ascending=False)
print(mediana_categoria)
```

**Digite aqui as categorias:**

Comparando média e mediana por categoria, temos:

| Categoria | Média | Mediana | Média x Mediana |
|---|---|---|---|
| comidas-preparadas | 3095,04 | 3290,00 | Média **abaixo** da mediana |
| belleza-y-cuidado-personal | 1783,56 | 1569,00 | Média **acima** da mediana |
| congelados | 2108,04 | 1519,00 | Média **acima** da mediana |
| frutas | 1724,47 | 1195,00 | Média **acima** da mediana |
| instantaneos-y-sopas | 765,49 | 439,00 | Média **acima** da mediana |
| lacteos | 2385,22 | 989,00 | Média **acima** da mediana |
| verduras | 1343,30 | 1180,00 | Média **acima** da mediana |

A única categoria em que a **média fica abaixo da mediana** é `comidas-preparadas`, o que sugere uma distribuição com cauda à esquerda (alguns produtos bem mais baratos puxando a média para baixo).

Em todas as demais categorias a **média está acima da mediana**, indicando distribuições assimétricas à direita (poucos produtos com preços bem altos puxando a média para cima). O caso mais extremo é `lacteos`, onde a média (2385,22) é mais que o **dobro** da mediana (989,00), sinal de forte assimetria e provável presença de outliers de preço alto.

# 2 - Traga o desvio padrão por categoria de produto.
# Qual o comportamento da média e mediana nas categorias com maior desvio?

```python
desvio_categoria = df.groupby('Categoria')['Preco_Normal'].std().sort_values(ascending=False)
print(desvio_categoria)
```

**Comportamento identificado:**

Ordenando por desvio padrão, do maior para o menor:

1. `lacteos` — desvio ≈ 3925,82
2. `belleza-y-cuidado-personal` — desvio ≈ 2210,04
3. `congelados` — desvio ≈ 2111,54
4. `comidas-preparadas` — desvio ≈ 2019,91
5. `frutas` — desvio ≈ 1639,15
6. `instantaneos-y-sopas` — desvio ≈ 1170,23
7. `verduras` — desvio ≈ 1012,70

A categoria `lacteos` tem, de longe, o maior desvio padrão. É justamente nela que observamos a maior distância entre média (2385,22) e mediana (989,00): a média é bem maior que a mediana, o que confirma que existe grande dispersão de preços com forte assimetria positiva (poucos produtos muito caros, como packs grandes de leite, puxando tanto a média quanto o desvio padrão para cima).

De forma geral, quanto maior o desvio padrão da categoria, maior tende a ser a diferença entre média e mediana, com a média ficando acima — isso é esperado, pois um desvio alto normalmente é causado por outliers de valor elevado, que distorcem a média sem afetar tanto a mediana.

# 3 - Plot um boxplot da distribuição do Preco_Normal para a categoria que você identificou que tem o maior desvio padrão. Como é a distribuição desses dados segundo o boxplot? Você identifica muitos outliers?

Dica: Para trazer apenas os dados da categoria que você deseja você pode usar o df.loc[df['Categoria'] == 'CATEGORIA ESCOLHIDA'

```python
lacteos_df = df.loc[df['Categoria'] == 'lacteos']

plt.figure(figsize=(8,6))
plt.boxplot(lacteos_df['Preco_Normal'], vert=True)
plt.title('Boxplot - Preco_Normal (Categoria: lacteos)')
plt.ylabel('Preco_Normal')
plt.show()
```

**Análise do boxplot (categoria `lacteos`):**

A distribuição do `Preco_Normal` para a categoria `lacteos` é fortemente assimétrica à direita: a caixa (que representa o intervalo entre o 1º e o 3º quartil) fica concentrada em valores mais baixos, próxima da mediana em torno de 989, enquanto a cauda superior se estende bastante.

Sim, é possível identificar **muitos outliers** acima do limite superior do boxplot (bigode superior), representando produtos como packs grandes/proteicos de leite com preços bem mais altos que a maioria dos produtos da categoria. Isso é coerente com o alto desvio padrão (≈3925,82) e com a diferença grande entre média (2385,22) e mediana (989,00) encontrada nas questões anteriores.

# 4 - Plote um gráfico de barras onde temos a média de descontos por categoria.

```python
desconto_media_categoria = df.groupby('Categoria')['Desconto'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,6))
desconto_media_categoria.plot(kind='bar', color='steelblue')
plt.title('Média de Desconto por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Média de Desconto')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# 5 - Plote um gráfico de mapa interativo agrupando os dados por categoria, marca e trazendo a média de desconto.

```python
desconto_categoria_marca = df.groupby(['Categoria', 'Marca'])['Desconto'].mean().reset_index()

fig = px.treemap(
    desconto_categoria_marca,
    path=['Categoria', 'Marca'],
    values='Desconto',
    color='Desconto',
    color_continuous_scale='RdYlGn_r',
    title='Média de Desconto por Categoria e Marca'
)
fig.show()
```
